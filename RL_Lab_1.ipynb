**Reinforcement Learning Laboratory - Experiment 1 _Exploring Reinforcement Learning Environments using Gymnasium**

```python
import sys
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import gymnasium as gym

def section_header(title: str):
    print("\n" + "=" * 70)
    print(f"  {title.upper()}")
    print("=" * 70)
```

**Task 1: Environment Setup**

```python
# =========================================================================
# Task 1: Environment Setup
# =========================================================================
section_header("Task 1: Environment Setup")
print("[+] Libraries imported successfully:")
print(f"    - Gymnasium version : {gym.__version__}")
print(f"    - NumPy version     : {np.__version__}")
print(f"    - Matplotlib version: {matplotlib.__version__}")
print("[+] Status: Environment Setup Complete.\n")
```

**Task 2: Create and Initialize an RL Environment**

```python
# =========================================================================
# Task 2: Create and Initialize an RL Environment
# =========================================================================
section_header("Task 2: Create and Initialize an RL Environment")
env_name = "CartPole-v1"
env = gym.make(env_name)
print(f"[+] Environment '{env_name}' created successfully.")

# Reset the environment (seed set for reproducible output)
initial_observation, info = env.reset(seed=42)

print(f"[+] Initial Observation (Reset output): {initial_observation}")
print(f"[+] Initial Info Dictionary          : {info}")
print(f"[+] Environment Spec                 : {env.spec.id}")
print(f"[+] Max Episode Steps                : {env.spec.max_episode_steps}")
```

**Task 3: Explore Observation and Action Spaces**

```python
# =========================================================================
# Task 3: Explore Observation and Action Spaces
# =========================================================================
section_header("Task 3: Explore Observation and Action Spaces")

obs_space = env.observation_space
act_space = env.action_space

print(f"[+] Observation Space             : {obs_space}")
print(f"[+] Observation Space Type        : {type(obs_space).__name__}")
print(f"[+] Observation Space Shape       : {obs_space.shape}")
print(f"[+] Observation High Bounds       : {obs_space.high}")
print(f"[+] Observation Low Bounds        : {obs_space.low}")
print("-" * 50)
print(f"[+] Action Space                  : {act_space}")
print(f"[+] Action Space Type             : {type(act_space).__name__}")
print(f"[+] Number of Possible Actions (n): {act_space.n}")
print("    - Action 0: Push cart to the LEFT")
print("    - Action 1: Push cart to the RIGHT")
```

**Task 4: Execute a Random Agent**

```python
# =========================================================================
# Task 4: Execute a Random Agent
# =========================================================================
section_header("Task 4: Execute a Random Agent")
print("Running one complete episode using a Random Agent...\n")

observation, info = env.reset(seed=42)
terminated = False
truncated = False
step_count = 0
cumulative_reward = 0.0

# Header format for step logs
print(f"{'Step':<6} | {'Action':<8} | {'Reward':<8} | {'Terminated':<10} | {'Truncated':<10} | {'Observation (Cart Pos, Vel, Angle, Pole Vel)':<50}")
print("-" * 105)

while not (terminated or truncated):
  step_count += 1

  # Select random action from action space
  action = env.action_space.sample()

  # Execute action in environment
  obs, reward, terminated, truncated, info = env.step(action)

  cumulative_reward += reward
  obs_str = np.array2string(obs, precision=4, suppress_small=True)

  print(f"{step_count:<6} | {action:<8} | {reward:<8.1f} | {str(terminated):<10} | {str(truncated):<10} | {obs_str:<50}")

env.close()

# Summary of Episode Execution
section_header("Task 4 Execution Summary")
print(f"[+] Total Steps Executed : {step_count}")
print(f"[+] Cumulative Reward    : {cumulative_reward}")
print(f"[+] Termination Status   : Terminated={terminated}, Truncated={truncated}")
print("=" * 70)
print("  EXPERIMENT 1 COMPLETED SUCCESSFULLY!")
print("=" * 70 + "\n")
```
